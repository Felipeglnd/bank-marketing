```python
import pandas as pd
import numpy as np
```

```python
caminho = r'..\data\raw\bank-additional-full.csv'
df = pd.read_csv(caminho, sep=';')
```

```python
# Visualizar todas as colunas
pd.set_option('display.max_columns', None)
df.head()
```

```python
df.shape
```

```python
df = df.drop('duration', axis=1)
```

```python
df['pdays'] = df['pdays'].replace(999, -1)
```

```python
df['flg_contacted_prev'] = np.where(df['pdays'] != -1, 1, 0)
print(df[['pdays', 'flg_contacted_prev']].head(10))
```

```python
df['total_credits_active'] = np.where(df['housing'] == 'yes', 1, 0) \
                                + np.where(df['loan'] == 'yes', 1, 0)
```

```python
df['total_contacts'] = np.where(df['previous'] > 0, df['previous'], 0) \
                            + np.where(df['campaign'] > 0, df['campaign'], 0)
```

```python
df['ratio_campaign_previous'] = round(df['campaign'] / np.where(df['previous'] > 0, df['previous'], 1), 2)
```

```python
df['economic_stress_index'] = df['euribor3m'] * df['emp.var.rate']
```

```python
map_meses = {'jan':1, 'feb':2, 'mar':3, 'apr':4, 'may':5, 'jun':6,
             'jul':7, 'aug':8, 'sep':9, 'oct':10, 'nov':11, 'dec':12
}

df['month_int'] = df['month'].map(map_meses)

max_meses = df['month_int'].max()

df['months_since_campaign'] = max_meses - df['month_int']

# Validação da transformação
print(df[['month', 'month_int', 'months_since_campaign']].sample(10))
```
