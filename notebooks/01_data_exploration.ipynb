# 1. Entendimento do Negócio (Business Understanding)
1. Contexto do Problema
O banco realizou campanhas de marketing direto (ligações telefônicas) para oferecer um produto financeiro: o Depósito a Prazo (Term Deposit).

2. Objetivo de Negócio
Objetivo Principal: Aumentar a taxa de conversão da campanha de marketing bancário, identificando quais clientes têm maior propensão a assinar o depósito a prazo.

Objetivo Secundário (Eficiência Operacional): Reduzir os custos de telemarketing, evitando ligar para clientes com baixíssima probabilidade de conversão.

3. Variável Target
y: Indica se o cliente contratou o depósito a prazo ('yes' ou 'no').

## Perguntas de Negócio a ser respondidas:

- Perfil Sociodemográfico: Qual faixa etária, profissão, nível educacional e estado civil apresentam a maior taxa de conversão?

- Histórico Financeiro/Risco: Clientes com empréstimo habitacional (housing), empréstimo pessoal (loan) ou histórico de inadimplência (default) têm menor disposição para investir em depósitos a prazo?

- Esforço de Campanha: Existe um ponto de saturação no número de contatos (campaign) a partir do qual insistir na ligação reduz a chance de conversão?

- Recência e Histórico de Campanhas Anteriores: Clientes que já foram contactados em campanhas passadas (pdays, previous, poutcome) têm maior propensão de conversão?

- Contexto Macroeconômico: Como variáveis de mercado (taxa Euribor a 3 meses, índice de confiança do consumidor, variação no emprego) influenciam a tomada de decisão do cliente?

```python
def visualizar_metadados(df) -> pd.DataFrame:
    """
    Função para visualizar Metadados de um Dataframe.
    return: Tipo, Cardinalidade, Qtd de Nulos e Percent Nulos
    param: Dataframe fornecido para análise
    """

    metadata = pd.DataFrame({
        'nome_variavel': df.columns,
        'tipo': df.dtypes,
        'qtd_nulos': df.isnull().sum(),
        'percent_nulos': round(df.isnull().sum() / len(df) * 100, 2),
        'cardinalidade': df.nunique()
    })

    metadata = metadata.sort_values(ascending=False, by='cardinalidade')
    metadata = metadata.reset_index(drop=True)

    return metadata
```

## Análise de Outliers (Bivariada)

```python
def boxplots(df, target_col='y', ncols_grid=3):
    """
    Gera boxplots bivariados para todas as variáveis numéricas do DataFrame
    em relação à variávei target.
    """

    # Identifica colunas numéricas desconsiderando o target
    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    if target_col in num_cols:
        num_cols.remove(target_col)

    # Validação das variáveis
    n_vars = len(num_cols)
    if n_vars == 0:
        print('Nenhuma variável numérica encontrada para plotar.')
        return

    # Calcular a quantidade necessárias de linha do grid
    nrows_grid= int(np.ceil(n_vars / ncols_grid))

    fig, axes = plt.subplots(
        nrows=nrows_grid, ncols=ncols_grid, figsize=(5 * ncols_grid, 4 * nrows_grid)
    )

    # Garante que o 'axes' seja tratável como lista unidimensional
    if n_vars == 1:
        axes = np.array([axes])
    axes = axes.flatten()

    # Mapeando as cores
    palettle = {'no': '#e74c3c', 'yes': '#2ecc71', 0: '#e74c3c', 1: '#2ecc71'}

    # Iteração para gerar os gráficos
    for i, col in enumerate(num_cols):
        sns.boxplot(
            data=df,
            x=target_col,
            y=col,
            ax=axes[i],
            palette=palettle,
            hue=target_col,
            legend=False
        )

        axes[i].set_title(
            f'Boxplot: {col} vs {target_col}',
            fontsize=11,
            fontweight='bold'
        )
        axes[i].set_xlabel(f'Target ({target_col})')
        axes[i].set_ylabel(col)

    # Remove Subplots vazios
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()
```

```python
boxplots(df, target_col='y', ncols_grid=3)
```

## Análise Kde Plots (Bivariada)

```python
def kde_plots(df, target_col='y', ncols_grid=3):
    """
    Gerar gráfico de densidade (KDE plots) bivariados para todas as variáveis
    numéricas do DataFrame em relação a variável target.
    """

    # Identifica a coluna numérica desconsiderando a target
    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    if target_col in num_cols:
        num_cols.remove(target_col)

    n_vars = len(num_cols)
    if n_vars == 0:
        print('Nenhuma variável numérica encontrada para plotar')
        return

    # Calcula linhas necessárias no grid
    nrows_grid = int(np.ceil(n_vars / ncols_grid))

    fig, axes = plt.subplots(
        nrows= nrows_grid, ncols= ncols_grid, figsize= (5 * ncols_grid, 3.8 * nrows_grid)
    )

    if n_vars == 1:
        axes = np.array([axes])
    axes = axes.flatten()

    # Mapeamento de cores (funciona para target 0/1 ou 'no'/'yes')
    palette = {'no': '#e74c3c', 'yes': '#2ecc71', 0: '#e74c3c', 1: '#2ecc71'}

    # Iteração para gerar cada KDE plot
    for i, col in enumerate(num_cols):
        sns.kdeplot(
            data=df,
            x=col,
            hue=target_col,
            common_norm=False, # Normaliza cada curva independentemente do desbalaceamento
            fill=True,
            alpha=0.35,
            palette=palette,
            ax=axes[i],
            linewidth=1.5
        )

        axes[i].set_title(
            f'Densidade: {col} por {target_col}',
            fontsize=11,
            fontweight='bold'
        )

        axes[i].set_ylabel('Densidade')
        axes[i].set_xlabel(col)

    # Remove os subplots vazios
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()
```

```python
kde_plots(df, target_col='y', ncols_grid=3)
```

## Análise das Variáveis Categóricas (Univariada)

```python
def analises_categoricas(df, cutoff=25, n_colsgrid=3):
    """
    Gerar gráfico de barras horinzontais para variáveis categóricas
    param df: Dataframe para ser selecionado as variáveis categóricas
    param cutoff = Limite máximo de categorias únicas permitidas para plotar
    param n_colsgrid = Quantidade de colunas no grid de gráficos
    """

    # Filtrar variáveis categórias
    cols_categoricas = df.select_dtypes(include=(['object', 'string'])).columns.tolist()

    # Filtrar pelo cutoff
    cols_categoricas = [col for col in cols_categoricas if df[col].nunique() <= cutoff]

    n_vars = len(cols_categoricas)
    if n_vars == 0:
        print(f'Não existe colunas categóricas no Dataframe')
        return

    # Calcula linhas do grid
    n_rows= int(np.ceil(n_vars / n_colsgrid))

    fig, axes = plt.subplots(
        nrows= n_rows, ncols=n_colsgrid, figsize=(7 * n_colsgrid, 3.8 * n_rows)
    )

    # Garante que axes seja uma lista unidimensional iterável
    if n_vars == 1:
        axes = np.array([axes])
    axes = axes.flatten()

    for i, col in enumerate(cols_categoricas):
        sns.countplot(
            data=df,
            y=col,
            ax=axes[i],
            color= 'royalblue',
            order=df[col].value_counts().index # Ordena do maior para o mennor
        )

        axes[i].set_title(f'Análise da Variável: {col}', fontsize=12, fontweight='bold')
        axes[i].set_ylabel(col, fontweight='bold')
        axes[i].set_xlabel('Quantidade')

    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()
```

```python
analises_categoricas(df, cutoff=25, n_colsgrid=3)
```

## Análise de Variáveis Categóricas (Bivariada)

```python
def analise_biraviada_categorica(df, target_col='y', cutoff=25, n_colsgrid=3):
    """
    Função para gerar gráficos de barras empilhados 100% para variáveis categóricas em relação ao TARGET
    """

    # Selecionando colunas categóricas (excluindo o target)
    cols_categoricas = df.select_dtypes(include=['object', 'string']).columns.tolist()
    if target_col in cols_categoricas:
        cols_categoricas.remove(target_col)

    # Selecionando variáveis pelo cutoff
    cols_categoricas = [col for col in cols_categoricas if df[col].nunique() <= cutoff]

    n_vars = len(cols_categoricas)

    if n_vars == 0:
        print(f'Não existe colunas categóricas no DataFrame')
        return

    # Qtd de linhas dos subplots
    n_rows = int(np.ceil(n_vars / n_colsgrid))

    fig, axes = plt.subplots(
        nrows=n_rows, ncols=n_colsgrid, figsize=(6 * n_colsgrid, 5 * n_rows)
    )

    if n_vars == 1:
        axes = np.array([axes])
    axes = axes.flatten()

    # Padronização das cores
    color_no = '#e74c3c'  # Vermelho
    color_yes = '#2ecc71' # Verde

    # Geração dos gráficos de barras empilhadas 100%:
    for i, col in enumerate(cols_categoricas):
        cross_tab = pd.crosstab(df[col], df[target_col], normalize='index') * 100 # Calcula a proporção

        if 'yes' in cross_tab.columns:
            cross_tab = cross_tab.sort_values(by='yes', ascending=True)
        elif 1 in cross_tab.columns:
            cross_tab = cross_tab.sort_values(by=1, ascending=True)

        # Plota a barra empilhada
        cross_tab.plot(
            kind='barh',
            stacked=True,
            ax=axes[i],
            color=[color_no, color_yes] if cross_tab.columns[0] in ['no', 0] else [color_yes, color_no],
            edgecolor='white',
            legend='False'
        )

        # Adicionar os rótulos de % dentro das barras
        for container in axes[i].containers:
            axes[i].bar_label(container, fmt='%.1f%%', label_type='center', color='black', fontsize=9, fontweight='bold')

        axes[i].set_title(f'Conversão (%) por {col}', fontsize=12, fontweight='bold')
        axes[i].set_ylabel(col)
        axes[i].set_xlabel('Taxa de Conversão (%)')
        axes[i].set_xlim(0, 100)

    # Remove subplots vazios e ajusta legenda]
    for j in range(i + 1, len(axes)):
        fig.delaxes(axes[j])

    plt.tight_layout()
    plt.show()
```

```python
analise_biraviada_categorica(df, target_col='y', cutoff=25, n_colsgrid=3)
```

### Insights Bivariados: Eficiência de Conversão (Stacked Bar Plots)

Ao analisar a proporção de conversão (`yes` vs `no`) dentro de cada categoria, descobrimos que os maiores volumes de esforço do banco não se traduziram em vendas reais. O cruzamento revelou os verdadeiros nichos de alta propensão.

### 1. O Paradoxo do Volume vs. Eficiência
* **A Ilusão de Maio (`month = may`):** Apesar de concentrar quase 1/3 de todas as chamadas da base, o mês de maio apresenta a **pior taxa de conversão (6,4%)**. Isso sugere uma campanha massiva e mal segmentada, gerando ineficiência operacional.
* **O Perfil `blue-collar`:** É a segunda profissão mais contatada (aprox. 9 mil ligações), mas possui a pior taxa de conversão entre os empregos **(apenas 6,9%)**.

### 2. Perfis de Alta Propensão (Nichos Lucrativos)
* **Extremos da Vida Financeira:** As maiores taxas de adesão estão entre **Estudantes (31,4%)** e **Aposentados (25,2%)**. O produto tem excelente fit com públicos que buscam guardar dinheiro a longo prazo (jovens iniciando a vida ou idosos protegendo patrimônio).
* **Sazonalidade Assertiva:** Os meses de **março, setembro, outubro e dezembro** tiveram baixíssimo volume de contatos, mas taxas de conversão excepcionais **(variando de 43,9% a 50,5%)**.

### 3. Comportamento e Histórico
* **O Peso do Histórico (`poutcome`):** Clientes que aceitaram ofertas em campanhas passadas (`success`) possuem uma impressionante taxa de conversão de **65,1%** na campanha atual. Esta será, sem dúvida, uma das variáveis de maior peso no modelo preditivo.
* **Canal de Comunicação:** Abordagens via celular (`cellular` - 14,7%) são quase 3x mais eficientes que telefones fixos (`telephone` - 5,2%).

# 4 - Conclusão da Análise Exploratória (Business Insights)

Com base na exploração univariada, bivariada e nas matrizes de correlação, respondemos às principais perguntas de negócio que guiarão a construção do modelo preditivo:

### 1. Perfil Sociodemográfico
**Pergunta:** *Qual faixa etária, profissão, nível educacional e estado civil apresentam a maior taxa de conversão?*
* **Insight:** A propensão à conversão é mais forte nos extremos da vida financeira. Clientes categorizados como estudantes (jovens) e aposentados (idosos) apresentam as maiores taxas de adesão a depósitos a prazo, destoando significativamente do público adulto padrão. Em contrapartida, profissionais da classe operária (`blue-collar`) possuem a pior eficiência de conversão, apesar de receberem um alto volume de ligações.

### 2. Histórico Financeiro e Risco
**Pergunta:** *Clientes com empréstimo habitacional (housing), empréstimo pessoal (loan) ou histórico de inadimplência (default) têm menor disposição para investir?*
* **Insight:** Sim. A base revelou que a imensa maioria dos clientes não possui empréstimos pessoais ativos (`loan = no`). Em relação à inadimplência (`default`), praticamente não há clientes com histórico confirmado de calote na base; a divisão ocorre entre quem "não deve" e quem o banco "não sabe" (`unknown`). A ausência dessa informação de crédito (`unknown`) atua como um forte sinal preditivo de risco financeiro.

### 3. Esforço de Campanha
**Pergunta:** *Existe um ponto de saturação no número de contatos (campaign) a partir do qual insistir na ligação reduz a chance de conversão?*
* **Insight:** Absolutamente. Os dados mostram que a grande maioria das conversões ocorre já nos primeiros contatos (mediana de 1 a 2 ligações). Ultrapassar o limite de 3 a 4 chamadas para o mesmo cliente na mesma campanha resulta em desgaste e desperdício de esforço operacional, com taxa de sucesso quase nula em *outliers* que chegam a dezenas de tentativas.

### 4. Recência e Histórico de Campanhas Anteriores
**Pergunta:** *Clientes que já foram contactados em campanhas passadas (pdays, previous, poutcome) têm maior propensão de conversão?*
* **Insight:** Este é o **fator preditivo mais forte da base**. O sucesso em uma campanha anterior (`poutcome = success`) gera uma taxa de conversão altíssima (acima de 65%). Além disso, como a maioria da base nunca foi contactada (`pdays = 999`), qualquer cliente com recência de contato inferior a 30 dias se destaca drasticamente na probabilidade de aceite.

### 5. Contexto Macroeconômico
**Pergunta:** *Como variáveis de mercado influenciam a tomada de decisão do cliente?*
* **Insight:** O impacto é estrutural e sistêmico. Existe uma fortíssima colinearidade entre a variação de empregos (`emp.var.rate`), o índice de preços e a taxa de juros Euribor (`euribor3m`). Momentos de taxas de juros mais baixas concentram os maiores picos de conversão. Taxas de juros altas, especialmente acompanhadas de retração no emprego, costumam diminuir significativamente a propensão dos clientes a prenderem dinheiro em investimentos a prazo.
